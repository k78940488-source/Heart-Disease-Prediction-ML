```python
import csv

with open("heart.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]

print(columns)
print("Rows:", len(data))
```

```python
values = []

age_index = columns.index("Age")

for row in data:
    if row[age_index] != "":
        values.append(float(row[age_index]))

print(values[:10])
```

```python
total = 0

for value in values:
    total = total + value

mean = total / len(values)

print("Mean:", mean)
```

```python
sorted_values = values.copy()
sorted_values.sort()

n = len(sorted_values)
middle = n // 2

if n % 2 == 0:
    median = (
        sorted_values[middle - 1]
        + sorted_values[middle]
    ) / 2
else:
    median = sorted_values[middle]

print("Median:", median)
```

```python
frequency = {}

for value in values:

    if value in frequency:
        frequency[value] = frequency[value] + 1
    else:
        frequency[value] = 1

mode = values[0]
highest_count = 0

for value in frequency:

    if frequency[value] > highest_count:
        highest_count = frequency[value]
        mode = value

print("Mode:", mode)
```

```python
minimum = values[0]

for value in values:

    if value < minimum:
        minimum = value

print("Minimum:", minimum)
```

```python
maximum = values[0]

for value in values:

    if value > maximum:
        maximum = value

print("Maximum:", maximum)
```

```python
total_squared_difference = 0

for value in values:

    difference = value - mean

    squared_difference = difference * difference

    total_squared_difference = (
        total_squared_difference
        + squared_difference
    )

variance = total_squared_difference / len(values)

standard_deviation = variance ** 0.5

print("Standard Deviation:", standard_deviation)
```

```python
numeric_columns = [
    "Age",
    "RestingBP",
    "Cholesterol",
    "FastingBS",
    "MaxHR",
    "Oldpeak",
    "HeartDisease"
]

results = {}

for column in numeric_columns:

    column_index = columns.index(column)

    values = []

    for row in data:

        if row[column_index] != "":
            values.append(float(row[column_index]))

    # Mean
    total = 0

    for value in values:
        total = total + value

    mean = total / len(values)

    # Median
    sorted_values = values.copy()
    sorted_values.sort()

    n = len(sorted_values)
    middle = n // 2

    if n % 2 == 0:
        median = (
            sorted_values[middle - 1]
            + sorted_values[middle]
        ) / 2
    else:
        median = sorted_values[middle]

    # Mode
    frequency = {}

    for value in values:

        if value in frequency:
            frequency[value] = frequency[value] + 1
        else:
            frequency[value] = 1

    mode = values[0]
    highest_count = 0

    for value in frequency:

        if frequency[value] > highest_count:
            highest_count = frequency[value]
            mode = value

    # Minimum
    minimum = values[0]

    for value in values:

        if value < minimum:
            minimum = value

    # Maximum
    maximum = values[0]

    for value in values:

        if value > maximum:
            maximum = value

    # Standard Deviation
    total_squared_difference = 0

    for value in values:

        difference = value - mean
        squared_difference = difference * difference

        total_squared_difference = (
            total_squared_difference
            + squared_difference
        )

    variance = (
        total_squared_difference
        / len(values)
    )

    standard_deviation = variance ** 0.5

    results[column] = {
        "Mean": mean,
        "Median": median,
        "Mode": mode,
        "Minimum": minimum,
        "Maximum": maximum,
        "Standard Deviation": standard_deviation
    }
```

```python
for column in results:

    print("\nColumn:", column)
    print("-------------------------")

    print("Mean:", results[column]["Mean"])
    print("Median:", results[column]["Median"])
    print("Mode:", results[column]["Mode"])
    print("Minimum:", results[column]["Minimum"])
    print("Maximum:", results[column]["Maximum"])
    print(
        "Standard Deviation:",
        results[column]["Standard Deviation"]
    )
```
