```python
import pandas as pd

df = pd.read_csv("loan_data_cleaned(2).csv")
```

```python
import numpy as np
import pandas as pd

cols = ["applicantincome", "coapplicantincome", "loanamount"]


winsor_df = df.copy()
log_df = df.copy()

for col in cols:

    
    low = df[col].quantile(0.01)
    high = df[col].quantile(0.99)

    winsor_df[col] = df[col].clip(low, high)

    
    log_df[col] = np.log1p(df[col].clip(lower=0))

print("Outlier Handling Completed!")
```
