```python
import csv
import io
```

```python
with open("heart.csv", "r", encoding="utf-8") as file:
    file_text = file.read()
```

```python
lines = file_text.splitlines()

print(lines[0])
```

```python
first_line = lines[0]

if "," in first_line:
    delimiter = ","

elif ";" in first_line:
    delimiter = ";"

elif "\t" in first_line:
    delimiter = "\t"

elif "|" in first_line:
    delimiter = "|"

else:
    delimiter = None

print("Delimiter:", repr(delimiter))
```

```python
reader = csv.reader(
    io.StringIO(file_text),
    delimiter=delimiter
)

rows = list(reader)

print(rows[:5])
```

```python
columns = rows[0]

data = rows[1:]

print("Columns:")
print(columns)

print("Total Rows:", len(data))
```

```python
print(data[0])
```

```python
print(columns)
```

```python
sort_column = "Age"

column_index = columns.index(sort_column)

print("Column Index:", column_index)
```

```python
data.sort(
    key=lambda row: float(row[column_index])
)
```

```python
print(data[:5])
```

```python
data.sort(
    key=lambda row: float(row[column_index]),
    reverse=True
)

print(data[:5])
```

```python
rows_per_page = 10

page_number = 1

start = (page_number - 1) * rows_per_page

end = start + rows_per_page

page_data = data[start:end]

print(page_data)
```

```python
page_number = 2

start = (page_number - 1) * rows_per_page

end = start + rows_per_page

page_data = data[start:end]

print(page_data)
```
