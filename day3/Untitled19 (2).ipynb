# Perform one-hot encoding manually by implementing a custom encoding function that dynamically adapts to new unseen categories in future data.

```python
import csv

with open("heart.csv","r",encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]
```

# Manual One-Hot Encoding

```python
cp_index = columns.index("ChestPainType")

categories = []

for row in data:

    value = row[cp_index]

    if value not in categories:
        categories.append(value)

print(categories)
```

# One-Hot Encoding Function

```python
def one_hot_encode(value, categories):

    encoded = []

    for category in categories:

        if value == category:
            encoded.append(1)
        else:
            encoded.append(0)

    return encoded
```

# Encode Dataset

```python
encoded_rows = []

for row in data:

    cp_vector = one_hot_encode(
        row[cp_index],
        categories
    )

    new_row = row.copy()

    new_row.pop(cp_index)

    for number in cp_vector:
        new_row.append(number)

    encoded_rows.append(new_row)
```

# New Column Names

```python
new_columns = columns.copy()

new_columns.pop(cp_index)

for category in categories:

    new_columns.append(
        "CP_" + category
    )

print(new_columns)
```

# Check Result

```python
print("Before")
print(data[0])

print("\nAfter One-Hot Encoding")
print(encoded_rows[0])
```

# Dynamic Unknown Category

```python
def dynamic_one_hot(value, categories):

    if value not in categories:
        categories.append(value)

    return one_hot_encode(value, categories)
```
