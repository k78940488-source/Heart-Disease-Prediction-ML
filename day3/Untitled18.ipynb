# Check for duplicate rows using a custom comparison algorithm, remove them manually, and analyze how duplicates impact data integrity and machine learning models.

```python
import csv

with open("heart.csv","r",encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]
```

```python
age_index = columns.index("Age")

age_values = []

for row in data:
    age_values.append(float(row[age_index]))
```

# Min & Max Manually

```python
minimum = age_values[0]
maximum = age_values[0]

for value in age_values:

    if value < minimum:
        minimum = value

    if value > maximum:
        maximum = value

print("Min :", minimum)
print("Max :", maximum)
```

# Mean & Std Manually

```python
total = 0

for value in age_values:
    total += value

mean = total / len(age_values)

square_sum = 0

for value in age_values:
    square_sum += (value - mean) ** 2

variance = square_sum / len(age_values)

std = variance ** 0.5

print(mean)
print(std)
```

# Standardization

```python
standardized_age = []

for value in age_values:

    z = (value - mean) / std

    standardized_age.append(z)

print(standardized_age[:10])
```
