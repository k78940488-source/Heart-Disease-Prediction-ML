# Create a new feature based on chol and thalach (e.g., "risk_factor") using a complex formula incorporating domain knowledge, interaction terms, and percentile-based ranking.

```python
import csv

with open("heart.csv","r",encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]
```

# Column Index

```python
chol_index = columns.index("Cholesterol")
hr_index = columns.index("MaxHR")
```

# Percentile Function (Manual)

```python
chol_values = []

for row in data:
    chol_values.append(float(row[chol_index]))

sorted_chol = chol_values.copy()
sorted_chol.sort()

p75_index = int(0.75 * len(sorted_chol))
chol_p75 = sorted_chol[p75_index]

print(chol_p75)
```

# Risk Formula

```python
risk_factor = []

for row in data:

    chol = float(row[chol_index])
    hr = float(row[hr_index])

    score = (
        chol * 0.6
    ) + (
        (220 - hr) * 0.4
    )

    if chol > chol_p75:
        score = score + 20

    risk_factor.append(score)
```

# Add New Feature

```python
new_columns = columns.copy()

new_columns.append("Risk_Factor")

new_data = []

for i in range(len(data)):

    row = data[i].copy()

    row.append(risk_factor[i])

    new_data.append(row)
```
