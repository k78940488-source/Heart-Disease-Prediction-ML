```python
missing_values = {}

for i in range(len(columns)):

    column_name = columns[i]

    count = 0

    for row in data:

        if i >= len(row) or row[i].strip() == "":
            count = count + 1

    missing_values[column_name] = count


for column in missing_values:

    print(
        column,
        "->",
        missing_values[column],
        "missing values"
    )
```

```python
for column in missing_values:

    if missing_values[column] > 0:

        print(
            column,
            "->",
            missing_values[column]
        )
```

```python
total_rows = len(data)

for column in missing_values:

    count = missing_values[column]

    percentage = (
        count / total_rows
    ) * 100

    if count > 0:

        print(
            column,
            "->",
            round(percentage, 2),
            "%"
        )
```

```python
column_name = "Cholesterol"

column_index = columns.index(column_name)

values = []

for row in data:

    if row[column_index].strip() != "":
        values.append(float(row[column_index]))
```

```python
total = 0

for value in values:
    total = total + value

mean = total / len(values)

print("Mean:", mean)
```

```python
for row in data:

    if row[column_index].strip() == "":
        row[column_index] = str(mean)
```

```python
column_name = "ChestPainType"

column_index = columns.index(column_name)

frequency = {}

for row in data:

    value = row[column_index].strip()

    if value != "":

        if value in frequency:
            frequency[value] = frequency[value] + 1
        else:
            frequency[value] = 1
```

```python
mode = None
highest_count = 0

for value in frequency:

    if frequency[value] > highest_count:

        highest_count = frequency[value]
        mode = value

print("Mode:", mode)
```

```python
for row in data:

    if row[column_index].strip() == "":
        row[column_index] = mode
```
