# Convert categorical columns (sex, cp, thal, etc.) into numerical form without using built-in encoding functions, manually mapping categories and handling unknown values dynamically.

```python
import csv

with open("heart.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]

print("Columns:", columns)
```

# Categorical Columns

```python
categorical_columns = [
    "Sex",
    "ChestPainType",
    "RestingECG",
    "ExerciseAngina",
    "ST_Slope"
]

print(categorical_columns)
```

# Categories manually check

```python
category_values = {}

for column in categorical_columns:

    column_index = columns.index(column)

    unique_values = []

    for row in data:

        value = row[column_index].strip()

        if value not in unique_values:
            unique_values.append(value)

    category_values[column] = unique_values


for column in category_values:

    print("\nColumn:", column)
    print("Categories:", category_values[column])
```

# Manual Mapping

```python
mappings = {

    "Sex": {
        "M": 0,
        "F": 1
    },

    "ChestPainType": {
        "ATA": 0,
        "NAP": 1,
        "ASY": 2,
        "TA": 3
    },

    "RestingECG": {
        "Normal": 0,
        "ST": 1,
        "LVH": 2
    },

    "ExerciseAngina": {
        "N": 0,
        "Y": 1
    },

    "ST_Slope": {
        "Up": 0,
        "Flat": 1,
        "Down": 2
    }
}
```

# Unknown Values Dynamically Handle

```python
def encode_value(value, mapping):

    if value in mapping:

        return mapping[value]

    else:

        new_code = len(mapping)

        mapping[value] = new_code

        return new_code
```

```python
test_mapping = {
    "M": 0,
    "F": 1
}

print(encode_value("M", test_mapping))
print(encode_value("F", test_mapping))
print(encode_value("X", test_mapping))

print(test_mapping)
```

# Dataset Encode

```python
encoded_data = []

for row in data:

    new_row = row.copy()

    for column in categorical_columns:

        column_index = columns.index(column)

        value = new_row[column_index].strip()

        mapping = mappings[column]

        new_row[column_index] = encode_value(
            value,
            mapping
        )
```

# Before/After

```python
print("Original Row:")
print(data[0])

print("\nEncoded Row:")
print(encoded_data[0])
```
