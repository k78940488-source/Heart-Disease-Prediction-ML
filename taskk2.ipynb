# Visualize the distribution of all numerical columns using only raw Matplotlib commands, ensuring custom aesthetics, annotations, and multiple comparative plots for deeper insights.

```python
import matplotlib.pyplot as plt
```

```python
import csv

with open("heart.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]

print(columns)
print("Rows:", len(data))
```

# Numerical columns

```python
numeric_columns = [
    "Age",
    "RestingBP",
    "Cholesterol",
    "FastingBS",
    "MaxHR",
    "Oldpeak",
    "HeartDisease"
]

print(numeric_columns)
```

# Numerical data manually

```python
numeric_data = {}

for column in numeric_columns:

    column_index = columns.index(column)

    values = []

    for row in data:

        if row[column_index].strip() != "":
            values.append(float(row[column_index]))

    numeric_data[column] = values

print(numeric_data["Age"][:10])
```

# Age Distribution

```python
plt.figure(figsize=(10, 6))

plt.hist(
    numeric_data["Age"],
    bins=15,
    edgecolor="black"
)

plt.title(
    "Age Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Age",
    fontsize=12
)

plt.ylabel(
    "Frequency",
    fontsize=12
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()
```

# Numerical Columns Distribution

```python
for column in numeric_columns:

    plt.figure(figsize=(10, 6))

    plt.hist(
        numeric_data[column],
        bins=15,
        edgecolor="black"
    )

    plt.title(
        column + " Distribution",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel(
        column,
        fontsize=12
    )

    plt.ylabel(
        "Frequency",
        fontsize=12
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()
```

# Annotation add

```python
values = numeric_data["Age"]

counts, bins, patches = plt.hist(
    values,
    bins=15,
    edgecolor="black"
)

max_index = 0

for i in range(len(counts)):

    if counts[i] > counts[max_index]:
        max_index = i

max_frequency = counts[max_index]

bin_center = (
    bins[max_index]
    + bins[max_index + 1]
) / 2

plt.annotate(
    "Highest Frequency",
    xy=(bin_center, max_frequency),
    xytext=(bin_center + 5, max_frequency + 10),
    arrowprops=dict(arrowstyle="->"),
    fontsize=11
)

plt.title(
    "Age Distribution with Annotation",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Age")
plt.ylabel("Frequency")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()
```
