```python
import pandas as pd
import numpy as np

# 1. Load Dataset
df = pd.read_csv("heart.csv")

# 2. Separate Features and Target
X = df.drop(columns=["HeartDisease"])
y = df["HeartDisease"].values

# 3. Convert Categorical Data
X = pd.get_dummies(X, drop_first=True).astype(float).values

# 4. Train-Test Split
np.random.seed(42)
idx = np.random.permutation(len(y))
split = int(0.8 * len(y))

train, test = idx[:split], idx[split:]
X_train, X_test = X[train], X[test]
y_train, y_test = y[train], y[test]

# 5. Scale Data
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
std[std == 0] = 1

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

# 6. Sigmoid Function
def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500)))

# 7. Initialize Parameters
w = np.zeros(X_train.shape[1])
b = 0
lr = 0.01

# 8. Gradient Descent
for i in range(3000):
    p = sigmoid(X_train @ w + b)
    error = p - y_train

    w -= lr * (X_train.T @ error) / len(y_train)
    b -= lr * error.mean()

print("Model Training Completed!")

# 9. Prediction
pred = (sigmoid(X_test @ w + b) >= 0.5).astype(int)

# 10. Manual Evaluation
TP = np.sum((y_test == 1) & (pred == 1))
TN = np.sum((y_test == 0) & (pred == 0))
FP = np.sum((y_test == 0) & (pred == 1))
FN = np.sum((y_test == 1) & (pred == 0))

accuracy = (TP + TN) / len(y_test)
precision = TP / (TP + FP) if TP + FP else 0
recall = TP / (TP + FN) if TP + FN else 0

f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0

print("\nModel Evaluation")
print("Accuracy :", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall   :", round(recall * 100, 2), "%")
print("F1 Score :", round(f1 * 100, 2), "%")
```
