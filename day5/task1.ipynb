```python
# Display bar plots for categorical variables without using high-level plotting libraries, dynamically adjusting bar widths and colors for better interpretation.
```

```python
import matplotlib.pyplot as plt
```

```python
import csv

with open("heart.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]

print("Total rows:", len(data))
print(columns)
```

# Categorical columns

```python
categorical_columns = [
    "Sex",
    "ChestPainType",
    "RestingECG",
    "ExerciseAngina",
    "ST_Slope"
]

print(categorical_columns)
```

# Category frequency manually

```python
column_name = "Sex"

column_index = columns.index(column_name)

frequency = {}
```

```python
for row in data:

    value = row[column_index].strip()

    if value in frequency:

        frequency[value] = frequency[value] + 1

    else:

        frequency[value] = 1

print(frequency)
```

```python
categories = []
counts = []

for category in frequency:

    categories.append(category)
    counts.append(frequency[category])

print(categories)
print(counts)
```

# Bar width dynamically calculate

```python
number_of_categories = len(categories)

bar_width = 0.8

if number_of_categories > 5:

    bar_width = 0.6

elif number_of_categories <= 2:

    bar_width = 0.5

print("Bar width:", bar_width)
```

# Dynamic colors

```python
colors = []

for i in range(number_of_categories):

    if i % 2 == 0:

        colors.append("steelblue")

    else:

        colors.append("orange")

print(colors)
```

# Bar Plot

```python
x_positions = []

for i in range(number_of_categories):

    x_positions.append(i)
```

```python
plt.figure(figsize=(8, 6))

plt.bar(
    x_positions,
    counts,
    width=bar_width,
    color=colors,
    edgecolor="black"
)

plt.xticks(
    x_positions,
    categories
)

plt.title(
    "Sex Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sex")
plt.ylabel("Frequency")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()
```

# categorical columns

```python
for column_name in categorical_columns:

    column_index = columns.index(column_name)

    frequency = {}

    # Manual frequency calculation
    for row in data:

        value = row[column_index].strip()

        if value in frequency:

            frequency[value] += 1

        else:

            frequency[value] = 1

    categories = []
    counts = []

    for category in frequency:

        categories.append(category)
        counts.append(frequency[category])

    number_of_categories = len(categories)

    # Dynamic bar width
    if number_of_categories > 5:

        bar_width = 0.6

    elif number_of_categories <= 2:

        bar_width = 0.5

    else:

        bar_width = 0.7

    # Dynamic colors
    colors = []

    for i in range(number_of_categories):

        if i % 2 == 0:
            colors.append("steelblue")
        else:
            colors.append("orange")

    x_positions = []

    for i in range(number_of_categories):

        x_positions.append(i)

    # Plot
    plt.figure(figsize=(10, 6))

    plt.bar(
        x_positions,
        counts,
        width=bar_width,
        color=colors,
        edgecolor="black"
    )

    plt.xticks(
        x_positions,
        categories
    )

    plt.title(
        column_name + " Distribution",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel(column_name)
    plt.ylabel("Frequency")

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()
```

```python
for column_name in categorical_columns:

    column_index = columns.index(column_name)

    frequency = {}

    for row in data:

        value = row[column_index].strip()

        if value in frequency:

            frequency[value] += 1

        else:

            frequency[value] = 1

    categories = []
    counts = []

    for category in frequency:

        categories.append(category)
        counts.append(frequency[category])

    x_positions = []

    for i in range(len(categories)):

        x_positions.append(i)

    colors = []

    for i in range(len(categories)):

        if i % 2 == 0:
            colors.append("steelblue")
        else:
            colors.append("orange")

    plt.figure(figsize=(10, 6))

    plt.bar(
        x_positions,
        counts,
        width=0.7,
        color=colors,
        edgecolor="black"
    )

    plt.xticks(
        x_positions,
        categories
    )

    # Annotation
    for i in range(len(counts)):

        plt.text(
            x_positions[i],
            counts[i] + 5,
            str(counts[i]),
            ha="center",
            fontweight="bold"
        )

    plt.title(
        column_name + " Distribution",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel(column_name)
    plt.ylabel("Frequency")

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()
```
