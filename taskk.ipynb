# Check for duplicate rows using a custom comparison algorithm, remove them manually, and analyze how duplicates impact data integrity and machine learning models.


```python
import csv

with open("heart.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    rows = list(reader)

columns = rows[0]
data = rows[1:]

print("Columns:", columns)
print("Total Rows:", len(data))
```

# Custom comparison function

```python
def rows_are_equal(row1, row2):

    if len(row1) != len(row2):
        return False

    for i in range(len(row1)):

        if row1[i].strip() != row2[i].strip():
            return False

    return True
```

# identyfy Duplicate rows

```python
duplicate_indexes = []

for i in range(len(data)):

    for j in range(i + 1, len(data)):

        if rows_are_equal(data[i], data[j]):

            duplicate_indexes.append(j)

print("Duplicate Rows:", len(duplicate_indexes))
```

# Duplicate rows display

```python

print("Duplicate Row Indexes:")
print(duplicate_indexes)

for index in duplicate_indexes:

    print("\nDuplicate Row:")
    print(data[index])
```

# Duplicate rows manually remove

```python
cleaned_data = []

for i in range(len(data)):

    if i not in duplicate_indexes:

        cleaned_data.append(data[i])

print("Original Rows:", len(data))
print("Cleaned Rows:", len(cleaned_data))
```

# Duplicate removal verify

```python
remaining_duplicates = []

for i in range(len(cleaned_data)):

    for j in range(i + 1, len(cleaned_data)):

        if rows_are_equal(
            cleaned_data[i],
            cleaned_data[j]
        ):

            remaining_duplicates.append(j)

print(
    "Remaining Duplicates:",
    len(remaining_duplicates)
)
```

# Data Integrity Analysis
   analyze how duplicates impact data integrity

```python
original_rows = len(data)
duplicate_rows = len(duplicate_indexes)
cleaned_rows = len(cleaned_data)

duplicate_percentage = (
    duplicate_rows / original_rows
) * 100

print("Original Rows:", original_rows)
print("Duplicate Rows:", duplicate_rows)
print("Cleaned Rows:", cleaned_rows)
print(
    "Duplicate Percentage:",
    duplicate_percentage
)
```
